```python
import numpy as np
from scipy import stats

def frequentist_two_proportion_test(conv_a, n_a, conv_b, n_b):
    p_a = conv_a / n_a
    p_b = conv_b / n_b
    p_pool = (conv_a + conv_b) / (n_a + n_b)
    se = np.sqrt(p_pool * (1 - p_pool) * (1 / n_a + 1 / n_b))
    z = (p_b - p_a) / se
    p_value = 2 * (1 - stats.norm.cdf(abs(z)))
    return {"rate_a": p_a, "rate_b": p_b, "z_stat": z, "p_value": p_value}
```

```python
frequentist_two_proportion_test(120, 2000, 150, 2000)
```

```python
def bayesian_beta_binomial(conv_a, n_a, conv_b, n_b, prior_alpha=1, prior_beta=1,
                            n_samples=100_000):
    post_a = np.random.beta(prior_alpha + conv_a,
                             prior_beta + (n_a - conv_a),
                             n_samples)
    post_b = np.random.beta(prior_alpha + conv_b,
                             prior_beta + (n_b - conv_b),
                             n_samples)
    prob_b_better = np.mean(post_b > post_a)
    return {"post_a": post_a, "post_b": post_b, "prob_b_better": prob_b_better}
```

```python
result = bayesian_beta_binomial(120, 2000, 150, 2000)
result["prob_b_better"]
```
